```python
import os
import zipfile
import requests
import pandas as pd
import numpy as np

from pathlib import Path

# Create project directories
PROJECT_DIR = Path("ML_Assignment_2")
DATA_DIR = PROJECT_DIR / "dataset"

DATA_DIR.mkdir(parents=True, exist_ok=True)

print("Project directory:", PROJECT_DIR.resolve())
print("Dataset directory:", DATA_DIR.resolve())
```

```python
# Official UCI Adult dataset ZIP file
DATASET_URL = (
    "https://archive.ics.uci.edu/static/public/2/adult.zip"
)

ZIP_PATH = DATA_DIR / "adult.zip"

response = requests.get(DATASET_URL, timeout=60)
response.raise_for_status()

with open(ZIP_PATH, "wb") as file:
    file.write(response.content)

print("Dataset downloaded successfully.")
print("Downloaded size:", round(ZIP_PATH.stat().st_size / 1024, 2), "KB")
```

```python
with zipfile.ZipFile(ZIP_PATH, "r") as zip_ref:
    zip_ref.extractall(DATA_DIR)

print("Dataset extracted successfully.")

print("\nFiles available:")
for file in DATA_DIR.iterdir():
    print(file.name)
```

```python
columns = [
    "age",
    "workclass",
    "fnlwgt",
    "education",
    "education-num",
    "marital-status",
    "occupation",
    "relationship",
    "race",
    "sex",
    "capital-gain",
    "capital-loss",
    "hours-per-week",
    "native-country",
    "income"
]

print("Total columns:", len(columns))
print("Input features:", len(columns) - 1)
print("Target variable:", columns[-1])
```

```python
# Load original training data
train_df = pd.read_csv(
    DATA_DIR / "adult.data",
    names=columns,
    skipinitialspace=True,
    na_values="?"
)

# Load original test data
test_df = pd.read_csv(
    DATA_DIR / "adult.test",
    names=columns,
    skipinitialspace=True,
    na_values="?",
    comment="|"
)

# Remove trailing full stop from test target labels
test_df["income"] = test_df["income"].str.rstrip(".")

# Combine datasets for overall dataset inspection
df = pd.concat(
    [train_df, test_df],
    ignore_index=True
)

print("Training dataset shape:", train_df.shape)
print("Testing dataset shape:", test_df.shape)
print("Combined dataset shape:", df.shape)
```

```python
import sklearn

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

print("Scikit-learn version:", sklearn.__version__)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

print("All preprocessing imports successful!")
```

```python
# Handle missing numerical values and scale numerical features
numerical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

print("Numerical preprocessing pipeline created.")
```

```python
# Handle missing categorical values and encode categories
categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

print("Categorical preprocessing pipeline created.")
```

```python
# Apply appropriate transformations to each feature group
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            numerical_transformer,
            numerical_features
        ),
        (
            "categorical",
            categorical_transformer,
            categorical_features
        )
    ],
    remainder="drop"
)

print("Complete preprocessing pipeline created.")
```

```python
# Fit preprocessing only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Apply the learned transformations to test data
X_test_processed = preprocessor.transform(X_test)

print("Preprocessing completed successfully.")

print("\nOriginal training shape:", X_train.shape)
print("Processed training shape:", X_train_processed.shape)

print("\nOriginal testing shape:", X_test.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
# Verify that processed data contains no missing or infinite values
print("Training data contains NaN:",
      np.isnan(X_train_processed).any())

print("Testing data contains NaN:",
      np.isnan(X_test_processed).any())

print("Training data contains infinite values:",
      np.isinf(X_train_processed).any())

print("Testing data contains infinite values:",
      np.isinf(X_test_processed).any())

print("\nTraining samples:", X_train_processed.shape[0])
print("Testing samples:", X_test_processed.shape[0])
print("Processed features:", X_train_processed.shape[1])
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.calibration import CalibratedClassifierCV

from sklearn.pipeline import Pipeline

import joblib
import time

print("All classifier libraries imported successfully.")
```

```python
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42,
        max_depth=15
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=5
    ),

    "Naive Bayes": GaussianNB(),

    "Support Vector Machine": CalibratedClassifierCV(
        estimator=SVC(
            kernel="rbf",
            random_state=42
        ),
        method="sigmoid",
        cv=3,
        ensemble=False
    )
}

print("Total classifiers:", len(models))

for model_name in models:
    print(model_name)
```

```python
model_pipelines = {}

for model_name, classifier in models.items():

    model_pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("classifier", classifier)
        ]
    )

    model_pipelines[model_name] = model_pipeline

print("Model pipelines created:", len(model_pipelines))
```

```python
trained_models = {}
training_times = {}

for model_name, pipeline in model_pipelines.items():

    print("\nTraining:", model_name)

    start_time = time.time()

    pipeline.fit(X_train, y_train)

    end_time = time.time()

    training_time = end_time - start_time

    trained_models[model_name] = pipeline
    training_times[model_name] = training_time

    print("Training completed.")
    print("Time taken:", round(training_time, 2), "seconds")

print("\nAll five models trained successfully.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report
)

import pandas as pd
import numpy as np

print("Evaluation libraries imported successfully.")
```

```python
evaluation_results = []
confusion_matrices = {}
classification_reports = {}

for model_name, model in trained_models.items():

    print("\nEvaluating:", model_name)

    # Predict class labels
    y_pred = model.predict(X_test)

    # Predict probability for positive class
    y_prob = model.predict_proba(X_test)[:, 1]

    # Calculate all six evaluation metrics
    accuracy = accuracy_score(y_test, y_pred)

    auc = roc_auc_score(
        (y_test == ">50K").astype(int),
        y_prob
    )

    precision = precision_score(
        y_test,
        y_pred,
        pos_label=">50K"
    )

    recall = recall_score(
        y_test,
        y_pred,
        pos_label=">50K"
    )

    f1 = f1_score(
        y_test,
        y_pred,
        pos_label=">50K"
    )

    mcc = matthews_corrcoef(y_test, y_pred)

    # Store results
    evaluation_results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "AUC": auc,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "MCC": mcc
    })

    # Store confusion matrix
    confusion_matrices[model_name] = confusion_matrix(
        y_test,
        y_pred,
        labels=["<=50K", ">50K"]
    )

    # Store classification report
    classification_reports[model_name] = classification_report(
        y_test,
        y_pred,
        zero_division=0
    )

    print("Evaluation completed.")

print("\nAll models evaluated successfully.")
```

```python
results_df = pd.DataFrame(evaluation_results)

# Round metrics for readability
metric_columns = [
    "Accuracy",
    "AUC",
    "Precision",
    "Recall",
    "F1 Score",
    "MCC"
]

results_df[metric_columns] = (
    results_df[metric_columns].round(4)
)

display(
    results_df.sort_values(
        by="F1 Score",
        ascending=False
    ).reset_index(drop=True)
)
```

```python
for model_name, matrix in confusion_matrices.items():

    print("\nConfusion Matrix:", model_name)

    matrix_df = pd.DataFrame(
        matrix,
        index=["Actual <=50K", "Actual >50K"],
        columns=["Predicted <=50K", "Predicted >50K"]
    )

    display(matrix_df)
```

```python
for model_name, report in classification_reports.items():

    print("\nClassification Report:", model_name)
    print(report)
```

```python
import os
import joblib

MODEL_DIR = PROJECT_DIR / "models"
RESULT_DIR = PROJECT_DIR / "results"

MODEL_DIR.mkdir(parents=True, exist_ok=True)
RESULT_DIR.mkdir(parents=True, exist_ok=True)

print("Model directory:", MODEL_DIR.resolve())
print("Results directory:", RESULT_DIR.resolve())
```

```python
# Define model filenames
model_filenames = {
    "Logistic Regression": "logistic_regression.joblib",
    "Decision Tree": "decision_tree.joblib",
    "KNN": "knn.joblib",
    "Naive Bayes": "naive_bayes.joblib",
    "Support Vector Machine": "svm.joblib"
}

# Save each complete trained pipeline
for model_name, trained_model in trained_models.items():

    filename = model_filenames[model_name]
    filepath = MODEL_DIR / filename

    joblib.dump(trained_model, filepath)

    print("Saved:", filename)

print("\nAll five trained models saved successfully.")
```

```python
# Save comparison metrics
results_df.to_csv(
    RESULT_DIR / "model_comparison.csv",
    index=False
)

# Save confusion matrices
for model_name, matrix in confusion_matrices.items():

    safe_name = model_filenames[model_name].replace(
        ".joblib", ""
    )

    matrix_df = pd.DataFrame(
        matrix,
        index=["Actual <=50K", "Actual >50K"],
        columns=["Predicted <=50K", "Predicted >50K"]
    )

    matrix_df.to_csv(
        RESULT_DIR / f"{safe_name}_confusion_matrix.csv"
    )

# Save classification reports
for model_name, report in classification_reports.items():

    safe_name = model_filenames[model_name].replace(
        ".joblib", ""
    )

    with open(
        RESULT_DIR / f"{safe_name}_classification_report.txt",
        "w"
    ) as file:
        file.write(report)

# Save training times
pd.DataFrame(
    list(training_times.items()),
    columns=["Model", "Training Time (seconds)"]
).to_csv(
    RESULT_DIR / "training_times.csv",
    index=False
)

print("All experiment results saved successfully.")
```
